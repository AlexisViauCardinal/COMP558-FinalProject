```python
import numpy as np
import cv2 as cv
from matplotlib import pyplot as plt
```

```python
# video_name = "videos/VIDEO-20250109-142924.mp4"
video_name = "videos/VIDEO-20250109-160653.mp4"
cap_video = cv.VideoCapture(video_name)
ret, img = cap_video.read()
```

```python
# Initiate ORB detector
orb = cv.ORB_create(nfeatures = 1000, edgeThreshold=10, patchSize = 10)
# orb = cv.ORB_create()
 
# find the keypoints with ORB
kp = orb.detect(img,None)
 
# compute the descriptors with ORB
kp, des = orb.compute(img, kp)

print(des.shape)
 
# draw only keypoints location,not size and orientation
img2 = cv.drawKeypoints(img, kp, None, color=(0,255,0), flags=0)
plt.imshow(img2)
plt.show()
```
