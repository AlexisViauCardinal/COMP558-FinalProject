```python
import numpy as np
import cv2 as cv
from matplotlib import pyplot as plt
```

```python
video_name = "videos/VIDEO-20250109-142924.mp4"
cap_video = cv.VideoCapture(video_name)
ret, img = cap_video.read()
gray = cv.cvtColor(img,cv.COLOR_BGR2GRAY)
```

```python
corners = cv.goodFeaturesToTrack(gray, 1000, 0.001, 5)
corners = np.int_(corners)
 
for i in corners:
    x,y = i.ravel()
    cv.circle(img,(x,y),3,255,-1)
 
plt.imshow(img)
plt.show()
```
